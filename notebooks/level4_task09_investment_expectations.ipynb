# Task 9: Expectations from Investments

**Objective:** Summarize the expectations respondents have from their investments.

**Steps:**
1. Review the expectations column.
2. List and describe the common expectations.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/Data_set_2.csv")
```

```python
order = ["10%-20%", "20%-30%", "30%-40%"]

counts = df["Expect"].value_counts().reindex(order)
percent = (counts / len(df) * 100).round(1)

pd.DataFrame({"count": counts, "percent": percent})
```

```python
plt.figure(figsize=(6, 4))
bars = plt.bar(counts.index, counts.values, color="#4C72B0")

for bar in bars:
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.3,
             int(bar.get_height()), ha="center")

plt.title("Expected Returns from Investments")
plt.xlabel("Expected return")
plt.ylabel("Number of Respondents")
plt.savefig("../outputs/charts/task09_expectations.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
pd.crosstab(df["Avenue"], df["Expect"])
```

```python
(pd.crosstab(df["Avenue"], df["Expect"], normalize="index")["20%-30%"] * 100).round(1)
```

## Finding

Expectations fall into three groups:

- **20%-30% returns:** 32 respondents (80.0%). This is the typical expectation.
- **30%-40% returns:** 5 respondents (12.5%). A small, very optimistic group.
- **10%-20% returns:** 3 respondents (7.5%). The most modest group.

All answers lie between 10% and 40%. The share expecting 20%-30% within each avenue group is: Equity 80%, Fixed Deposits 89%, Mutual Fund 83%, Public Provident Fund 33% (PPF has only 3 respondents, so that figure rests on a single person). Assuming these are yearly returns, the expectations are well above what safer options such as fixed deposits usually pay, so they look optimistic.
