# Task 2: Gender Distribution

**Objective:** Visualize the gender distribution in the dataset.

**Steps:**
1. Extract the gender column from the dataset.
2. Count each gender and plot the distribution.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/Data_set_2.csv")
```

```python
counts = df["gender"].value_counts()
percent = (counts / len(df) * 100).round(1)

print(counts)
print()
print(percent)
```

```python
plt.figure(figsize=(6, 4))
bars = plt.bar(counts.index, counts.values, color=["#4C72B0", "#DD8452"])

for bar in bars:
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.3,
             int(bar.get_height()), ha="center")

plt.title("Gender Distribution of Respondents")
plt.xlabel("Gender")
plt.ylabel("Number of Respondents")
plt.savefig("../outputs/charts/task02_gender_bar.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(5, 5))
plt.pie(counts.values, labels=counts.index, autopct="%1.1f%%",
        startangle=90, colors=["#4C72B0", "#DD8452"])
plt.title("Gender Share of Respondents")
plt.savefig("../outputs/charts/task02_gender_pie.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Finding

Of the 40 respondents, 25 are male (62.5%) and 15 are female (37.5%). There are about 1.7 male respondents for every female respondent.

The sample is small and skewed towards men, so it may not represent investors in general. If I compare genders in later tasks, the female group has only 15 people, so any difference would be weak evidence.
