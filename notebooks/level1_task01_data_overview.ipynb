## Finding

The dataset has 40 rows and 24 columns. It is clean: there are 0 missing values and 0 duplicate rows.

Of the 24 columns, 8 are numeric and 16 are text. Only `age` is a true numeric measure. The other 7 numeric columns (`Mutual_Funds`, `Equity_Market`, `Debentures`, `Government_Bonds`, `Fixed_Deposits`, `PPF`, `Gold`) are preference ranks from 1 to 7, so they should be read as rankings, not amounts.

The text columns hold survey answers such as gender, investment avenue, objectives, reasons and information sources. One column name, `Stock_Marktet`, has a spelling mistake in the original data.
