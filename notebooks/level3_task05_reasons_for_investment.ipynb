# Task 5: Reasons for Investment

**Objective:** Analyze and summarize the reasons for investment choices.

**Steps:**
1. Explore the reason columns.
2. Identify common themes and summarize them.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/Data_set_2.csv")
```

```python
reason_cols = ["Reason_Equity", "Reason_Mutual", "Reason_Bonds", "Reason_FD"]

for col in reason_cols:
    print(col)
    print(df[col].value_counts())
    print()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 7))

for ax, col in zip(axes.flatten(), reason_cols):
    counts = df[col].value_counts()
    ax.barh(counts.index, counts.values, color="#4C72B0")
    ax.invert_yaxis()
    ax.set_title(col.replace("Reason_", "Reasons for ") )
    ax.set_xlabel("Respondents")

plt.tight_layout()
plt.savefig("../outputs/charts/task05_reasons.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
all_reasons = pd.concat([df[col] for col in reason_cols]).value_counts()
all_reasons
```

## Finding

Reasons by investment type:

- **Equity:** Capital Appreciation (30), Dividend (8), Liquidity (2)
- **Mutual Funds:** Better Returns (24), Fund Diversification (13), Tax Benefits (3)
- **Bonds:** Assured Returns (26), Safe Investment (13), Tax Incentives (1)
- **Fixed Deposits:** Risk Free (19), Fixed Returns (18), High Interest Rates (3)

Taken together, the most common reasons across all four types are Capital Appreciation (30), Assured Returns (26), Better Returns (24), Risk Free (19). Returns drive the choice for equity and mutual funds (capital appreciation and better returns), while safety drives the choice for bonds and fixed deposits (assured returns, safe investment, risk free). Tax-related reasons are rare in every group.
